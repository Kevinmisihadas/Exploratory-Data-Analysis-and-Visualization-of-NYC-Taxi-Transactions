# Exploratory Data Analysis and Visualization of NYC Taxi Data

```python
# Importing Required Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

### Loading the Taxis Dataset

```python
df=sns.load_dataset('taxis')
df
```

```python
#Baic Data Exploration
print('Shape:',df.shape,'\n')
print('Data Type:\n',df.dtypes,'\n')
print('Columns:\n',df.columns,'\n')
```

### Handling Missing Values

```python
#Checking missing values in the dataset
missing=df.isnull().sum()
print(missing,'\n')
#The dataset contains no missing values in the numerical columns and the color column.
#However, payment has 44 missing values,pickup_zone and pickup_borough have 26 missing values each, and dropoff_zone and dropoff_borough have 45 missing values each.
print('Columns Containg Missing Values:\n',missing[missing>0])
```

```python
# Impute missing values
# Missing Numerical Columns
col_num=df.select_dtypes(include='number').columns
col_cat=df.select_dtypes(include='object').columns
for col in col_num:
    df[col]=df[col].fillna(df[col].median())
for col in col_cat:
    df[col]=df[col].fillna(df[col].mode()[0])
```

```python
print(df.isnull().sum())
```

```python
# Remove rows with missing values to maintain data integrity.
df.dropna(subset=['pickup', 'fare', 'distance', 'pickup_borough'])
```

### Visualizations using Matplotlib/Pandas Plot:

```python
df.sample(3)
```

```python
#The pickup column is converted into datetime format to analyze the fare distribution over time. 
#A line chart is plotted with pickup timestamps on the x-axis and fare on the y-axis to identify changes and trends in taxi fares over time.
df['pickup']=pd.to_datetime(df['pickup'])

# Now the line chart follows time from earliest to latest.
df = df.sort_values('pickup')

#Plot
plt.figure(figsize=(8, 5))
plt.plot(df['pickup'],df['fare'])
plt.title('Fare Over Time')
plt.xlabel('Pickup Time')
plt.ylabel('Fare')
plt.xticks(rotation=50)
plt.show()
```

```python
'''The data is grouped by pickup_borough to calculate the total fare collected from each borough.
A bar chart is plotted to compare the total fare across different pickup boroughs and identify which borough generates the highest fare revenue.'''
borough=df.groupby('pickup_borough')['fare'].sum()
#Plot
borough.plot(kind='bar')
plt.title('Total Fare by Borough')
plt.xlabel('Pickup Borough')
plt.ylabel('Total Fare')
plt.show()
```

```python
'''The data is grouped by payment method to calculate the number of trips made using each payment type (credit card, cash, etc.). 
A pie chart is plotted to visualize the distribution of trips across different payment methods and identify the most commonly used payment method.'''
payment=df['payment'].value_counts()

# Plot
plt.figure(figsize=(7, 7))

plt.pie(
    payment,
    labels=payment.index,
    startangle=90,
    autopct='%1.1f%%'

)

plt.title('Distribution of Trips by Payment Method')
plt.show()
```

```python
'''The distance column is used to visualize the distribution of taxi trip distances. 
A histogram is plotted with a customized number of bins to understand how frequently different distance ranges occur and identify the most common trip distances.'''
plt.figure(figsize=(8,5))
plt.hist(df['distance'],bins=20,edgecolor='red',color='black')
plt.show()
```

```python
'''
The data is grouped by pickup_borough to compare the distribution of tip amounts across different boroughs.
A box plot is plotted with pickup_borough as the categorical axis and tip as the numerical axis to identify the median tip, spread of tips, and potential outliers in each borough.
'''
plt.figure(figsize=(8,5))
df.boxplot(column='tip',by='pickup_borough')
plt.title('Tip Distribution by Pickup Borough')
plt.xlabel('Pickup Borough')
plt.ylabel('Tip Amount')
plt.show()
```

# Visualizations using Seaborn:

```python
# Create a count plot to visualize the number of trips in each pickup borough.
plt.figure(figsize=(8,5))
sns.countplot(data=df,x='pickup_borough')
plt.title('Number of Trips by Pickup Borough')
plt.xlabel('Pickup Borough')
plt.ylabel('Number of Trips')
plt.show()
```

```python
df.sample(2)
```

```python
# Plot a scatter plot showing the relationship between distance and fare. Color the points based on pickup borough.
plt.figure(figsize=(8,5))
sns.scatterplot(data=df,x='distance',y='fare',hue='pickup_borough')
plt.title('Relationship between distance and fare')
plt.xlabel('Distance')
plt.ylabel('fare')
plt.show()
```

```python
# Plot a heatmap to visualize the correlation between numerical variables such as distance, fare, tip, tolls and total.
numerical_data=df[['distance', 'fare', 'tip', 'tolls','total']]
# Correlation matrix
corr=numerical_data.corr()

#Plot
plt.figure(figsize=(8,5))
sns.heatmap(corr,annot=True,cmap='coolwarm')
plt.title('Correlation between numerical variables')
plt.show()
```

```python
# Create a pair plot to visualize pairwise relationships between distance, fare, tip and total. Color the data points according to pickup_zone.
# Select required columns
pair_data = df[['distance', 'fare', 'tip', 'total', 'pickup_zone']].copy()

# Select the five most frequent pickup zones
top_zones = pair_data['pickup_zone'].value_counts().head(5).index

pair_data = pair_data[pair_data['pickup_zone'].isin(top_zones)]

# Create pair plot
sns.pairplot(
    pair_data,
    hue='pickup_zone'
)

plt.show()
```

```python
# Plot a violin plot showing the distribution of fare for each payment method.
plt.figure(figsize=(9, 5))
sns.violinplot(
    data=df,
    x='payment',
    y='fare'
)
plt.title('Fare Distribution by Payment Method')
plt.xlabel('Payment Method')
plt.ylabel('Fare')
plt.show()
```
